Filtering dataframe (loc and Iloc)

```python
# Returns students with marks greater than 90
df[df["mark"] > 90]
```

```python
df[df["class"]=="Seven"]
```

```python
df[(df["gender"]=="male") & (df["class"]=="four")]
```

```python
df[(df["mark"]<90) & (df["gender"]=="female")]
```

```python
# average mark of the female student
df[df["gender"]=="female"]["mark"].mean()
```

```python
# OR condition uses pipe symbol
df[(df["mark"]<90) | (df["gender"]=="female")]
```

```python
df.loc[df["mark"]>90, :]
# : means all columns
```

```python
df.loc[(df["mark"]>80) & (df["class"]=="Five"), ["name", "mark"]]
```

Using Filter Funtction

```python
df.filter(items=['name', 'mark'])
```

```python
# Returns column name starting or containing 'ma'
df.filter(like='ma')
```

```python
 # Query method filters rows
 # filter method filters columns
 df.query("mark > 60")
```

Sorting Results
# normal sort is ascending
# reverse sort to descending, ascending = false

```python
# Sorting by names
df.sort_values(by="name")
```

```python
df.sort_values(by="mark", ascending=False)
```

```python
df.sort_values(by ="class", ascending=False)
```

```python
# return students who are class seven or female
df[(df["gender"]=="female") | (df["class"]=="Seven")].sort_values(by="mark")
```

```python
df.loc[:,:].sort_values(by= ["mark", "class"])
```
